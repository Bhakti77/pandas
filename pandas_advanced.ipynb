```python
#Reshaping data
#melt()
#melt() converts a wide DataFrame to long format — opposite of pivot table.
import pandas as pd
wide_df = pd.DataFrame({
    'customer': ['Aarav', 'Priya', 'Rohan'],
    'jan_sales': [1000, 2000, 1500],
    'feb_sales': [1200, 1800, 1700],
    'mar_sales': [900, 2200, 1600]
})
wide_df
```

```python
long_df = wide_df.melt(id_vars='customer',var_name='month',value_name='sales')
```

```python
long_df.groupby('month')['sales'].sum()   # total sales per month
#long_df.groupby('customer')['sales'].mean()  # avg sales per customer
```

```python
#pivot - opposite of melt, converts long form back to wide format
long_df.pivot(index='customer', columns='month', values='sales')
```

```python
#apply in dataframes
df = pd.DataFrame({
    'q1_sales': [1000, 2000, 1500],
    'q2_sales': [1200, 1800, 1700],
    'q3_sales': [900, 2200, 1600]
}, index=['Aarav', 'Priya', 'Rohan'])

df
```

```python
df.apply('sum')
```

```python
# row-wise — axis=1
df.apply('sum', axis=1)
```

```python
#Method chaining
result = (df
    .query('amount > 5000')
    .groupby('city')['amount']
    .sum()
    .sort_values(ascending=False)
    .reset_index()
)
result
#one pipeline instead of 4 diff results — much more readable.
```

```python
#query() — cleaner filtering
Instead of:
pythondf[df['amount'] > 5000]
Use:
pythondf.query('amount > 5000')
df.query('city == "Mumbai"')
df.query('amount > 5000 and city == "Mumbai"')
More readable — especially in chains.
```

```python
#Time Series

import pandas as pd
import numpy as np

dates = pd.date_range('2024-01-01', periods=365, freq='D')

ts = pd.DataFrame({
    'date':   dates,
    'sales':  np.random.randint(1000, 5000, 365),
    'orders': np.random.randint(10, 100, 365)
})

ts = ts.set_index('date')
ts.head()
```

```python
# monthly total sales
ts.resample('ME').sum()
```

```python
#weekly average
ts.resample('QE').mean()
```
