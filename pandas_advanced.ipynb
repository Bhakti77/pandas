```python
#Reshaping data
#melt()
#melt() converts a wide DataFrame to long format — opposite of pivot table.
import pandas as pd
wide_df = pd.DataFrame({
    'customer': ['Aarav', 'Priya', 'Rohan'],
    'jan_sales': [1000, 2000, 1500],
    'feb_sales': [1200, 1800, 1700],
    'mar_sales': [900, 2200, 1600]
})
wide_df
```

```python
long_df = wide_df.melt(id_vars='customer',var_name='month',value_name='sales')
```

```python
long_df.groupby('month')['sales'].sum()   # total sales per month
#long_df.groupby('customer')['sales'].mean()  # avg sales per customer
```

```python
#pivot - opposite of melt, converts long form back to wide format
long_df.pivot(index='customer', columns='month', values='sales')
```

```python
#apply in dataframes
df = pd.DataFrame({
    'q1_sales': [1000, 2000, 1500],
    'q2_sales': [1200, 1800, 1700],
    'q3_sales': [900, 2200, 1600]
}, index=['Aarav', 'Priya', 'Rohan'])

df
```
