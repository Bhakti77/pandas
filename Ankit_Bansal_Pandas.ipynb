```python
df_missing.fillna("Unknown") #Null will changed to Unknown

#forwardfill
#fills null with the previous valid value:
s = pd.Series([100, None, None, 200, None, 300])

s.ffill()
# 100, 100, 100, 200, 200, 300

#backfill
s.bfill()
#100,200,200,200,300,300

```

```python
#Date and time in Pandas
df.head()
```

```python
df['order_date'] = pd.to_datetime(df['order_date'],format = '%d-%m-%Y')
df['ship_date'] = pd.to_datetime(df['ship_date'],format = '%d-%m-%Y')
#changed to datetime because dtype was object, and thus date functions cannot be performed
df.head()
df.dtypes
```

```python
df['order_date'].dt.year #dt is ancessor here, like str, because one column makes it as series and series has no such date methods
df['order_date'].dt.month
df['order_date'].dt.day
df['order_date'].dt.day_name()
df['order_date'].dt.quarter
```

```python
df[df['order_date']> "2021-01-01"]
```

```python
#Aggregation

df['sales'].sum()
df['sales'].mean()
df['sales'].min()
df['sales'].max()
df['sales'].count()
df['region'].nunique()
# df['sales'].size
```

```python
#column wise aggregation

df.groupby('region')['sales'].sum()  #sum of sales by region

df.groupby('category')['order_id'].count()  #count of orders by category

df.groupby('region')['sales'].agg(["min","max"])  #multiple aggregations using agg

df.groupby('region').agg({'sales':'sum','profit':'mean'})  #multiple aggregations using agg and giving column names
```

```python
df['category'].value_counts()
```

```python
#Pivot table

pd.pivot_table(
    df,
    index="region",
    columns='category',
    aggfunc='sum',
    values='sales',
    fill_value=0
)
```

```python
#Joins

df_orders = pd.read_csv("/orders.csv")
df_returns = pd.read_csv("/returns.csv")
df_returns.rename(columns={'return_order_id':'order_id'},inplace=True)
```

```python
joined_df = pd.merge(left=df_orders,right=df_returns, on='order_id',how='left') #left, right,cross
joined_df=joined_df.fillna("Not Returned")
joined_df.head()
```

```python
joined_df.groupby('return_reason')['order_id'].count()
```

```python
#concat same as SQL union all not union, means duplicate rows will come in result
#pd.concat([df1, df2])              keeps duplicates — like UNION ALL
#pd.concat([df1, df2]).drop_duplicates()   removes duplicates — like UNION

pd.concat([df_orders,df_returns])

```

```python
#to_csv - write to new csv file to_excel, to json_etc.
joined_df.to_csv("/full_orders.csv",index=False)
```
