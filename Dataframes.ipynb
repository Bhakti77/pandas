```python
import pandas as pd

data = {
    'name':       ['Aarav', 'Priya', 'Rohan', 'Sneha', 'Karan',
                   'Nisha', 'Amit', 'Pooja', 'Vikram', 'Anjali'],
    'city':       ['Mumbai', 'Pune', 'Delhi', 'Bangalore', 'Mumbai',
                   'Chennai', 'Pune', 'Mumbai', 'Delhi', 'Bangalore'],
    'category':   ['Electronics', 'Appliances', 'Electronics', 'Fitness',
                   'Electronics', 'Appliances', 'Fitness', 'Electronics',
                   'Fitness', 'Appliances'],
    'sales':      [75000, 3500, 82000, 1200, 45000,
                   4500, 6000, 45000, 3200, 12000],
    'rating':     [4.5, 3.2, None, 5.0, 2.8,
                   None, 4.1, 3.9, None, 4.7]
}

df = pd.DataFrame(data)
print(df)
```

```python
#handling nulls
df.isnull()
df.isnull().sum()
```

```python
#drop nulls
df.dropna()
```

```python
df.shape #drop returns new dataframe so original was unchanged
```

```python
df['rating'].fillna(df['rating'].mean())

#In real DA work filling with mean or median is called imputation —
#a standard data cleaning technique.Because 0 is misleading — a rating of 0 doesn't exist in reality, it would skew your analysis.
#If someone calculates average rating, 0 would pull it down unfairly.
```

```python
df['rating'] = df['rating'].fillna(df['rating'].mean())
df.isnull().sum()
```

```python
#Aggregations

#basics
df['sales'].sum()    # total sales
df['sales'].mean()   # average sales
df['sales'].max()    # highest sale
df['sales'].min()    # lowest sale
df['sales'].count()  # number of values
```

```python
df['sales'].describe()
```

```python
#GROUPBY - most powerful aggregation
df.groupby('city')['sales'].sum()
```

```python
df.groupby('category')['sales'].sum() # avg sales per category
```

```python
df.groupby('city')['sales'].count() # number of orders per city
```

```python
#Sorting
df.sort_values(by='sales') #default ascending
df.sort_values('sales', ascending=False)
```
