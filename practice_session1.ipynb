```python
import pandas as pd

data = {
    'order_id':   [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    'customer':   ['Aarav', 'Priya', 'Rohan', 'Sneha', 'Karan',
                   'Nisha', 'Amit', 'Pooja', 'Vikram', 'Anjali'],
    'city':       ['Mumbai', 'Pune', 'Delhi', 'Mumbai', 'Bangalore',
                   'Delhi', 'Pune', 'Mumbai', 'Bangalore', 'Delhi'],
    'category':   ['Electronics', 'Appliances', 'Electronics', 'Fitness',
                   'Electronics', 'Appliances', 'Fitness', 'Electronics',
                   'Fitness', 'Appliances'],
    'amount':     [75000, 3500, 82000, 1200, 45000,
                   4500, 6000, 45000, 3200, 12000],
    'status':     ['delivered', 'delivered', 'pending', 'delivered', 'cancelled',
                   'delivered', 'pending', 'delivered', 'cancelled', 'delivered']
}

df = pd.DataFrame(data)
```

```python
#Find the total amount spent per city. Sort results from highest to lowest.
total_amount = df.groupby('city')['amount'].sum()
total_amount.sort_values(ascending=False)
```

```python
#Find all orders where:
#status is 'delivered'
#amount is greater than 10000

stats = df[(df['status'] == 'delivered') & (df['amount'] > 10000)]
display(len(stats))
```

```python
#Find the average amount per category, but only for delivered orders.

delivered_data = df[df['status']=='delivered'].groupby('category')['amount'].mean()
display(delivered_data)
```

```python
#Add a new column called 'discount' to the DataFrame:
# If amount > 50000 → discount is 10%
# If amount > 10000 → discount is 5%
# Otherwise → discount is 0%

df['discount'] = df['amount'].apply(lambda x: 0.1* x if x>50000
                          else 0.05*x if x>10000 else 0)
df
```

```python
#Add another column called 'final_amount':

#final_amount = amount - discount

df['final_amount'] = df['amount'] - df['discount']
df
```
