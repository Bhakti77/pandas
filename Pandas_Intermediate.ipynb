```python
import pandas as pd
import numpy as np

data = {
    'order_id':   [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    'customer':   ['Aarav', 'Priya', 'Rohan', 'Sneha', 'Karan',
                   'Nisha', 'Amit', 'Pooja', 'Vikram', 'Anjali'],
    'city':       ['Mumbai', 'Pune', 'Delhi', 'Mumbai', 'Bangalore',
                   'Delhi', None, 'Mumbai', 'Bangalore', 'Delhi'],
    'category':   ['Electronics', 'Appliances', 'Electronics', 'Fitness',
                   'Electronics', 'Appliances', 'Fitness', 'Electronics',
                   None, 'Appliances'],
    'amount':     [75000, 3500, 82000, 1200, 45000,
                   None, 6000, 45000, 3200, 12000],
    'status':     ['delivered', 'delivered', 'pending', None, 'cancelled',
                   'delivered', 'pending', 'delivered', 'cancelled', 'delivered']
}

df = pd.DataFrame(data)
df
```

```python
df.isnull().sum()
```

```python
#agg - multiple aggregation same time

df.groupby('city')['amount'].agg(
    total_sales = 'sum',
    avg_sales = 'mean',
    max_sales = 'max'
)
```

```python
#multiple column in agg - amount and order_id
df.groupby('city').agg(
    total_amount  = ('amount', 'sum'),
    avg_amount    = ('amount', 'mean'),
    order_count   = ('order_id', 'count')
)
```

```python
#transform
df['city_total'] = df.groupby('city')['amount'].transform('sum')
df[['customer', 'city', 'amount', 'city_total']]
```

```python
df['category_total'] = df.groupby('category')['amount'].transform('sum')
df['pct_of_category'] = df['amount'] / df['category_total'] * 100
df[['customer', 'category', 'amount', 'category_total', 'pct_of_category']]
```

```python
df['pct_of_category'] = df['amount'] / df['city_total'] * 100
df[['customer', 'city', 'amount', 'city_total', 'pct_of_city']]
```

```python
#Handling duplicates
import pandas as pd
import numpy as np

data = {
    'order_id':   [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    'customer':   ['Aarav', 'Priya', 'Rohan', 'Sneha', 'Karan',
                   'Nisha', 'Amit', 'Pooja', 'Vikram', 'Anjali'],
    'city':       ['Mumbai', 'Pune', 'Delhi', 'Mumbai', 'Bangalore',
                   'Delhi', None, 'Mumbai', 'Bangalore', 'Delhi'],
    'category':   ['Electronics', 'Appliances', 'Electronics', 'Fitness',
                   'Electronics', 'Appliances', 'Fitness', 'Electronics',
                   None, 'Appliances'],
    'amount':     [75000, 3500, 82000, 1200, 45000,
                   None, 6000, 45000, 3200, 12000],
    'status':     ['delivered', 'delivered', 'pending', None, 'cancelled',
                   'delivered', 'pending', 'delivered', 'cancelled', 'delivered']
}

df = pd.DataFrame(data)

# add duplicates
duplicate_rows = df.iloc[[0, 2]]
df = pd.concat([df, duplicate_rows], ignore_index=True)
df.shape
```

```python
df.duplicated().sum()
```

```python
df[df.duplicated()]
```

```python
df = df.drop_duplicates() #can pass column names as subset as well, it check full combination
df.shape
```

```python
#concat -Used to combine two or more DataFrames by stacking them on top of each other (or side by side).

df1 = pd.DataFrame({
    'order_id': [111, 112,113],
    'customer': ['Rahul', 'Meera','Vikrant'],
    'city':     ['Nagpur', 'Surat','Jaipur'],
    'amount':   [5000, 8000,12000]
})

df2 = pd.DataFrame({
    'order_id': [113, 114],
    'customer': ['Vikrant', 'Kavya'],
    'city':     ['Jaipur', 'Kolkata'],
    'amount':   [12000, 9000]
})
```

```python
#WINDOW FUNCTIONS
#rank() and rolling()

df['amount_rank'] = df['amount'].rank(ascending=False)
df=df.sort_values('amount_rank')
df[['customer', 'amount', 'amount_rank']]
```

```python
#rank within groups
df['rank_in_city'] = df.groupby('city')['amount'].rank(ascending=False)
df[['customer', 'city', 'amount', 'rank_in_city']]
```
